# Truncatable Primes - Problem 37

The number $3797$ has an interesting property.  Being prime itself, it is possible to continuously remove digits from left to right, and remain prime at each stage: $3797, 797, 97,$ and $7$.  Similarly we can work from right to left: $3797, 379, 37,$ and $3$.

Find the sum of the only eleven primes that are both truncatable from left to right and right to left.

Note: $2, 3, 5,$ and $7$ are not considered to be truncatable primes.

# My solution

Since a lot of the primes are going to be covered by a smaller truncation, keeping a cache of prior input values can save a lot of work.

After that, it's a matter of iterating over the primes (excluding those with 1 digit) and continuing until the count hits 11.

```python
from functools import lru_cache

@lru_cache(10**6)
def is_prime(n):
    if n == 1:return False
    for p in primes():
        if p * p > n:return True
        if n % p == 0:return False

def primes():
    yield 2
    n = 3
    while True:
        if is_prime(n):yield n
        n += 2
```

```python
@lru_cache(10**4)
def truncatable_left(n):
    if n < 10:return True
    next = int(str(n)[1:])
    if not is_prime(next):return False
    return truncatable_left(next)

@lru_cache(10**4)
def truncatable_right(n):
    if n < 10:return True
    next = int(str(n)[:-1])
    if not is_prime(next):return False
    return truncatable_right(next)

count = 0
total = 0
for p in primes():
    if p < 10:continue
    if truncatable_left(p) and truncatable_right(p):
        count += 1
        total += p
    if count == 11:break

total
```
