# Pandigital Multiples - Problem 38

Take the number $192$ and multiply it by each of $1, 2,$ and $3$:

$192 \times 1 = 192$

$192 \times 2 = 384$

$192 \times 3 = 576$

By concatenating each product we get the $1$ to $9$ pandigital, $192384576$.  We will call $192384576$ the concatenated product of $192$ and $(1, 2, 3)$.

The same can be achieved by starting with $9$ and multiplying by $1, 2, 3, 4,$ and $5$, giving the pandigital, $918273645$, which is the concatenated product of $9$ and $(1,2,3,4,5)$.

What is the largest $1$ to $9$ pandigital $9$-digit number that can be formed as the concatenated product of an integer with $(1,2,...,n)$ where $n \gt 1$?

# My solution

Given that $n$ has to be at least $2$, that means our $m$ (the value we multiply by) has to be at most $4$ digits

$1 < m \lt 10^5$

This is because, with an $m$ that's $5$ digits long, $m \times 1$ would be $5$ digits, and $m \times 2$ would be at least $5$ digits, meaning $10$ in total (more than our allowed $9$ digits).

That gives us our finite search space, now we just need to figure out how to check each value of $m$.

Another insight is that the first digit of $m$ must be a $9$.

Given that the solution of $m=9$ is provided as $918273645$, that means $m \times 1$ must start with a $9$, which is only possible if $m$ does too.

So given that we know $m = 9$ gives $918273645$ and is the best for 1 digit $m$, we can look for other possible ranges.

Since $m = 9$ produces a result that starts with a $9$, that means the only values of $m$ that can possibly improve are those which also start with a $9$.

|m Range|n Range|Valid|
|----|---|---|
|90 to 99|(90, 99), (180, 198), (270, 297), (360, 396)|None (Can't hit exactly 9 digits)|
|900 to 999|(900, 999), (1800, 1998), (2700, 2997)|None (Can't hit exactly 9 digits)|
|9000 to 9999|(9000, 9999), (18000, 18998)|All|

This means we only need to check $9000 \leq m \leq 9999$ for improvements on $m = 9$

```python
from itertools import permutations

largest_cprod = 918273645

#candidates = [i for i in range(9000, 10000) if len(str(i)) == len(set(str(i)))]
#candidates = [i for i in candidates if not "0" in str(i)]

candidates = [9000 + int("".join(i)) for i in permutations("12345678", r=3)]
candidates = [i for i in candidates if len(set(str(i) + str(2 * i))) == 9]
candidates = [i for i in candidates if not "0" in str(2 * i)]

print(candidates)

for m in candidates:

    cprod = int(str(m) + str(2 * m))
    if len(set(str(cprod))) != 9:continue
    if "0" in str(cprod):continue

    largest_cprod = max(largest_cprod, cprod)

largest_cprod
```
